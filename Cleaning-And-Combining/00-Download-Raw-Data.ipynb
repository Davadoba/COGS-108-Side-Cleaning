# Download the raw data

Run this once after cloning. It downloads the files that come from a public URL into `data-files/00-raw/` and **skips any file that is already there**.

The other raw files have to be added by hand (or by their own notebook):

| Raw file | Dataset | How to get it |
|---|---|---|
| `bbref_team_salaries_2015_2026.csv` | #1 | Scraped from Basketball-Reference team pages (`modules/scrape_team_salaries.js`) |
| `PlayerStatistics.csv` | #2 | Download from [Kaggle](https://www.kaggle.com/datasets/eoinamoore/historical-nba-data-and-player-box-scores) (needs a Kaggle account) |
| `NBA Salaries(1990-2023).csv` | #3 | Download from [Kaggle](https://www.kaggle.com/datasets/loganlauton/nba-players-and-team-data) |
| `nba_league_leaders_raw.csv` | #4 | Created automatically by `04-NBA-Stats-Player-Stats.ipynb` the first time it runs (calls the NBA Stats API) |

```python
import requests
from tqdm import tqdm

from nba_cleaning_utils import RAW

# Dataset #5 (Mr-Bridge NBA salary/cap dataset) is a set of public CSVs on Hugging Face.
HF_BASE = 'https://huggingface.co/datasets/Mr-Bridge/nba-salary-cap-contracts-2016-2026/resolve/main'
datafiles = [
    {'url': f'{HF_BASE}/player_salaries.csv', 'filename': 'hf_player_salaries.csv'},
    {'url': f'{HF_BASE}/team_payrolls.csv',   'filename': 'hf_team_payrolls.csv'},
    {'url': f'{HF_BASE}/cap_tax_aprons.csv',  'filename': 'hf_cap_tax_aprons.csv'},
    {'url': f'{HF_BASE}/min_salaries.csv',    'filename': 'hf_min_salaries.csv'},
    # Dataset #6: NBA.com's official player-movement (transactions) log, one public JSON file
    {'url': 'https://stats.nba.com/js/data/playermovement/NBA_Player_Movement.json',
     'filename': 'nba_player_movement.json'},
]

for f in tqdm(datafiles, desc='Downloading'):
    path = RAW / f['filename']
    if path.exists():
        print('already have', path)
        continue
    response = requests.get(f['url'], timeout=60)
    response.raise_for_status()
    path.write_bytes(response.content)
    print('downloaded', path)
```

```python
# Which raw files are present?
expected = ['bbref_team_salaries_2015_2026.csv', 'PlayerStatistics.csv', 'NBA Salaries(1990-2023).csv',
            'nba_league_leaders_raw.csv'] + [f['filename'] for f in datafiles]
for name in expected:
    print('OK     ' if (RAW / name).exists() else 'MISSING', name)
```
