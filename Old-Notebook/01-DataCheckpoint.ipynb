## Research Question

Among NBA players who are under contract for the following season, how well can a player's contract situation (salary as a percentage of the salary cap and years left on the contract), their team's payroll position (under the cap, over the cap, over the luxury tax, or over an apron), and their regular-season and playoff performance (minutes, points, and plus-minus per game) predict whether that player is traded in the offseason after each season from 2015–16 to 2024–25?

We define the offseason as the period from the end of a regular season to the opening night of the next season. Trades are frozen from the February trade deadline until then, so this captures every trade made between seasons. We measure prediction on the two most recent seasons, which the model never sees during training. We use the area under the precision-recall curve (PR-AUC) because trades are rare, and compare it against two baselines: a model that guesses using only the overall trade rate, and a model that uses regular-season statistics alone.

## Background and Prior Work

The National Basketball Association (NBA) is a professional basketball league that consists of 30 teams, where teams regularly make decisions about their rosters based on player contracts, performance, and financial considerations. Our group is interested in investigating whether certain factors can help predict if an NBA player will be traded during the offseason. ## Hypothesis

We hypothesize that a neural network trained on player-seasons from 2015–16 to 2022–23 can predict which players are traded in the offseason better than chance. We will test it on the 2023–24 and 2024–25 offseasons, which it never sees during training. We expect its PR-AUC on those seasons to be at least twice the share of players who were actually traded (about 0.06, so a PR-AUC of at least 0.12).

We expect the following features to have more importance:
1. Contract length: Players with one year left on their contract will be traded more often than players with three or more years left. An expiring contract is easier for the receiving team to take on, and it lets the trading team get something back before the player can leave in free agency.
2. Playoff performance: Players who played more playoff minutes, or had a higher playoff plus-minus, will be traded less often. Teams tend to keep players who contributed to postseason success.
3. Team payroll: Players on teams over the luxury tax or an apron will be traded more often than players on teams under the cap, especially when the player's salary is a large share of the cap. Those teams face financial penalties and roster restrictions that push them to cut salary.

We will check how true this is by comparing trade rates between groups. For example, the percentage traded that had one year left vs three years left on their contracts. We will determine which predictors are stronger by using permutation importance: shuffling different groups of variables in the test seasons and measuring how much the model's PR-AUC drops. A bigger drop means those variables are stronger predictors. We expect the contract variables and the playoff variables to each cause a larger drop than the regular-season statistics.

#### 3F. Clean the data

We make a cleaned copy and keep the raw `salaries` table unchanged. Here is each choice and why we made it:

1. **Remove the row with no player ID.** It's not a player and we can't match it to anything. It's 1 row out of 7,863.
2. **Remove exact duplicate rows.** These are identical copies, so they carry no new information.
3. **Turn a salary of 0 into a blank.** In step 3D, zeros (2018-2022) and blanks (2023-2026) looked like the same thing recorded two ways, so we use one value for both. We **don't drop** these rows, because the player was still on that team that season and we need to know who was on each roster to study trades. We **don't fill them** with 0 or an average, because we don't know the real salary and a made-up number would hurt the analysis. Instead we add a column, `no_salary_listed`, that marks them.
4. **Keep one row per player, team, and season.** When a player has 2 or more rows for the same team and season, we keep the largest salary. We aren't sure what the extra lines mean (a second short contract, or money still owed), so adding them could count the same money twice. The column `n_salary_lines` shows how many rows were combined.
5. **Use one spelling of each name.** Each `player_id` gets the most recent spelling of the name.
6. **Add two helper columns:** `very_small_salary` (under 100,000, from step 3E) and `season_label` (like `2014-15`, so it's clear which season the end-year number means).

```python
print('rows at the start:', len(salaries))

clean = salaries.copy()

# 1. remove the row with no player ID
clean = clean[clean['player_id'].notna()]
print('after removing the row with no player ID:', len(clean))

# 2. remove exact duplicate rows
clean = clean.drop_duplicates()
print('after removing exact duplicates:', len(clean))

# 3. a salary of 0 becomes a blank (NaN), so both kinds of "no salary" look the same
clean['salary'] = clean['salary'].mask(clean['salary'] == 0)

# 4. one row per player, team, and season (keep the largest salary)
n_lines = clean.groupby(key).size().rename('n_salary_lines')
clean = clean.groupby(key, as_index=False).agg(player=('player', 'last'), salary=('salary', 'max'))
clean = clean.merge(n_lines, on=key)
print('after one row per player, team, and season:', len(clean))

# 5. one spelling of each name (the most recent one)
latest_name = clean.sort_values('season').groupby('player_id')['player'].last()
clean['player'] = clean['player_id'].map(latest_name)

# 6. helper columns
clean['no_salary_listed'] = clean['salary'].isna()
clean['very_small_salary'] = clean['salary'] < 100000
clean['season_label'] = (clean['season'] - 1).astype(str) + '-' + (clean['season'] % 100).astype(str).str.zfill(2)

clean = clean[['season_label', 'season', 'team', 'player_id', 'player', 'salary',
               'n_salary_lines', 'no_salary_listed', 'very_small_salary']]
clean = clean.sort_values(['season', 'team', 'salary'], ascending=[True, True, False]).reset_index(drop=True)

clean.head(10)
```

```python
print('rows, columns:', clean.shape)
print('different players:', clean['player_id'].nunique())
print('rows that combined 2 or more lines:', (clean['n_salary_lines'] > 1).sum())
print('rows with no salary listed:', clean['no_salary_listed'].sum(),
      '(', round(100 * clean['no_salary_listed'].mean(), 1), '% )')
print('player IDs with more than one spelling of the name:', clean.groupby('player_id')['player'].nunique().gt(1).sum())

clean.isna().sum()
```

### Dataset #2: NBA Box Scores & Player Game Statistics (Kaggle) (#1)

#### Data Description & Important Metrics (2A)
This dataset contains historical individual game logs for NBA players spanning regular season, preseason, and playoff games. The observational unit is an individual player's performance within a single scheduled game. Key metrics include:
- **`numMinutes` (Playing Time):** Measured in continuous minutes (e.g., 27.82 minutes). A value of 0 or NaN indicates the player did not enter the game (DNP). Standard regulation games last 48 minutes (overtime adds 5 minutes per period), meaning typical individual playing times range from 5 to 40 minutes.
- **Counting Stats (`points`, `assists`, `reboundsTotal`, `steals`, `blocks`, `turnovers`):** Measured in integer counts per game. For typical rotation players, points range between 5 and 25 per game (rarely exceeding 50), while single-game assists and rebounds typically fall between 0 and 15.
- **Shooting Efficiency (`fieldGoalsPercentage`, `threePointersPercentage`, `freeThrowsPercentage`):** Expressed as decimal proportions bounded between 0.000 and 1.000. These represent shots made divided by shots attempted. A league-average true shooting percentage hovers around 55% to 58%, while elite three-point shooters typically shoot between 37% and 42% over a season.

#### Dataset Concerns and Limitations (2B)
1. **Survivorship & Playing Time Bias:** The sample reflects active roster selections on game nights. Bench players with sporadic playing time exhibit extreme variance in rate stats (e.g., going 1-for-1 from three equates to a 100% shooting percentage over 2 minutes), which can distort model feature weights if minimum minute thresholds are not applied.
2. **Missing Not at Random (MNAR) in Box Scores:** Inactive players (due to injury, load management, or coach decisions) register as NaN for minutes and counting metrics rather than 0. Dropping these rows ensures per-game averages reflect true on-court production, but requires aggregating row counts separately to track player durability (Games Played).
3. **Era Differences and Rule Shifts:** Spanning from 1947 to present, older eras lack tracking metrics (e.g., the three-point line was introduced in 1979; blocks, steals, and turnovers were not tracked prior to 1973). To ensure feature consistency for predictive trade modeling, we restrict our subset to the modern era (2015 onwards).

### Dataset #4: NBA Player Per-Game Stats, 2014-15 to 2025-26 Regular Seasons

- **Dataset name:** NBA Stats API, `leagueleaders` endpoint (per-game stats, regular season)
- **Link:** https://stats.nba.com/stats/leagueleaders (we followed the request format from the `pseudo-r/Public-NBA-API` project)
- **Saved as:** `data/00-raw/nba_league_leaders_raw.csv` (raw) and `data/02-processed/nba_player_stats_cleaned.csv` (cleaned) (collected 11/8/2026)
- **Number of observations:** 2,996 rows (one row per player per season, 850 different players, 12 seasons)
- **Number of variables:** 29 (25 from the API, plus 4 we added: `season_label`, `season`, `pts_per_min`, `pts_outlier`)
- **How we collected it:** We used Python's `requests` library to ask the NBA Stats API for the regular season of each of the 12 seasons from 2014-15 to 2025-26, sorted by points per game. The API only answers requests that include certain headers (`x-nba-stats-origin` and `x-nba-stats-token`), so we added them. We waited 3 seconds between requests, added a `SEASON` column to each answer, stacked the 12 tables, and saved the result without other changes in `data/00-raw/`. The API only returns players who qualify for its leaderboard, which matters for the concerns below. (We also tried the `leaguedashplayerstats` endpoint, which lists every player, but it did not respond to our scripted requests, so we kept the leaderboard.)

This dataset covers the on-court performance part of our research question. It tells us how much each player scored, rebounded, passed, and played in each regular season, which gives us the performance side to compare with contracts and offseason trades.

#### 2B. Concerns with this dataset

1. **Only regular season.** Playoff games have shorter rotations and tougher defense, so a good regular-season scorer may look different in the playoffs. Playoff results are not in this dataset and will come from another source.
2. **Only qualified players are included.** The API leaves out players with few games (the lowest `gp` is 45), so bench players and injured players are missing. These may be exactly the players most likely to be traded, so on its own this dataset would bias our results toward regular players. To cover them, we will take regular-season stats for players who are not on the leaderboard from Dataset #2 (Kaggle box scores), which has a game log for every player who appeared in a game. Both datasets use the NBA's player IDs (`player_id` here, `personId` in Dataset #2), so we can tell which players are missing here and fill them in from Dataset #2.
3. **The number of players changes by season.** There are between 228 and 271 players per season, and some teams have far fewer qualified players than others. This is not missing data, but it means teams and seasons are not equally represented.
4. **Some seasons were shorter.** The 2019-20 and 2020-21 seasons had fewer games, so `gp` is not comparable across seasons. Per-game averages like `pts` and `min` are still comparable.
5. **The same player appears in many seasons.** 850 players make up 2,996 rows, so rows from the same player are not independent. We need to keep this in mind when we build and test a model.
6. **One team per row.** Each row lists one `team`. Some rows have `gp` above 82, which suggests the stats of a player who played for more than one team are combined. We have not yet checked which team is listed for these players (see 3E).
7. **Scoring changes over the years.** Pace and 3-point shooting have changed since 2014-15, so we flag scoring outliers within each season and not across all seasons together.
8. **No contract or trade information.** There is no salary, contract length, or trade history, so this dataset covers only the performance side of our question.
9. **Minutes affect every stat.** Players who play more minutes pile up more points and rebounds, so we added `pts_per_min` to compare players on a fairer scale. A high value in few minutes can still be noise.

```python
import os
import time
import requests
import pandas as pd
import matplotlib.pyplot as plt

os.makedirs('data/00-raw', exist_ok=True)
os.makedirs('data/02-processed', exist_ok=True)

raw_path = 'data/00-raw/nba_league_leaders_raw.csv'
SEASONS = [f'{y}-{str(y + 1)[2:]}' for y in range(2014, 2026)]   # 2014-15 ... 2025-26

def fetch_season(season):
    headers = {
        "Accept": "application/json, text/plain, */*",
        "Origin": "https://www.nba.com",
        "Referer": "https://www.nba.com/",
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36",
        "x-nba-stats-origin": "stats",
        "x-nba-stats-token": "true",
    }
    params = {"LeagueID": "00", "PerMode": "PerGame", "Scope": "S",
              "Season": season, "SeasonType": "Regular Season", "StatCategory": "PTS"}
    response = requests.get("https://stats.nba.com/stats/leagueleaders",
                            headers=headers, params=params, timeout=30)
    response.raise_for_status()
    result_set = response.json()['resultSet']
    one = pd.DataFrame(result_set['rowSet'], columns=result_set['headers'])
    one.insert(0, 'SEASON', season)
    return one

if os.path.exists(raw_path):
    nba_raw = pd.read_csv(raw_path)
else:
    pieces = []
    for season in SEASONS:
        pieces.append(fetch_season(season))
        print('downloaded', season, pieces[-1].shape)
        time.sleep(3)          # be polite to the API
    nba_raw = pd.concat(pieces, ignore_index=True)
    nba_raw.to_csv(raw_path, index=False)

print('raw shape:', nba_raw.shape)
nba_raw.head()
```

#### 3E. Outliers and suspicious entries

We use the IQR rule: a value is an outlier if it is more than 1.5 × IQR below Q1 or above Q3 (IQR = Q3 − Q1). We use it because scoring is skewed (a few stars score far more than everyone else), so the rule does not rely on the average the way a Z-score does. We apply it **within each season**, because scoring changes from year to year and a single cutoff for all 12 seasons would flag more players in high-scoring seasons.

```python
def flag_iqr(series):
    q1, q3 = series.quantile(0.25), series.quantile(0.75)
    iqr = q3 - q1
    return (series < q1 - 1.5 * iqr) | (series > q3 + 1.5 * iqr)

nba['pts_outlier'] = nba.groupby('season_label')['pts'].transform(flag_iqr)
min_outlier = nba.groupby('season_label')['min'].transform(flag_iqr)

print('scoring outliers:', nba['pts_outlier'].sum(), ', minutes outliers:', min_outlier.sum())
print(nba.groupby('season_label')['pts_outlier'].sum())
nba[nba['pts_outlier']].nlargest(10, 'pts')[['season_label', 'player', 'team', 'pts', 'min', 'gp']]
```

```python
# Entries that don't make sense: percentages outside 0-1, more shots made than attempted, negative values
pct_cols = ['fg_pct', 'fg3_pct', 'ft_pct']
print('percentages outside 0-1:', ((nba[pct_cols] < 0) | (nba[pct_cols] > 1)).sum().sum())
print('made more than attempted:', ((nba['fgm'] > nba['fga']) | (nba['fg3m'] > nba['fg3a']) | (nba['ftm'] > nba['fta'])).sum())
print('negative values:', (nba.select_dtypes('number') < 0).sum().sum())

nba.nlargest(5, 'pts')[['player', 'team', 'pts', 'min', 'gp']]

# Games played above 82 (a team only plays 82 games)
print('rows with gp above 82:', (nba['gp'] > 82).sum())
nba[nba['gp'] > 82][['season_label', 'player', 'team', 'gp', 'min', 'pts']]
```

We think these are players who were traded during the season and whose games with both teams are added together, since two teams' schedules don't match on the day of the trade. This is our guess, because the data doesn't say. We keep these rows, and they may help us find players who changed teams in a season.

We make a cleaned copy and keep the raw table unchanged. Here is each choice and why we made it:

2. **Keep players with at least 10 games.** This is a safety check for players with too few games to trust. It removes 0 rows here, because everyone has at least 45 games.

4. **Keep `pts_outlier`.** The flag we made in 3E marks the 75 scoring outliers. We **don't drop** them because they are real players with real stats.

6. **Sort by season and rank**, with the identifying columns first.

- Remember that this table covers only players who qualify for the leaderboard (see 2B).

### Dataset #5: NBA Salary Cap & Contracts, 2015-16 to 2026-27 (Mr-Bridge, Hugging Face)

**Source and access.** [Mr-Bridge/nba-salary-cap-contracts-2016-2026](https://huggingface.co/datasets/Mr-Bridge/nba-salary-cap-contracts-2016-2026) is a free, public dataset of 15 CSV tables. Player salaries come from HoopsHype's public salary data (snapshot accessed 2026-07-07). League cap rules come from the 2017 and 2023 NBA/NBPA collective bargaining agreements and NBA cap releases, cross-checked against Basketball-Reference and Spotrac. We use four of the tables, downloaded by the setup cell above into `data/00-raw/`:
- `player_salaries.csv` (7,288 rows): one row per player, per team, per season. This is the core table.
- `cap_tax_aprons.csv` (12 rows): the league's salary cap, luxury-tax line, and first/second apron for each season.
- `min_salaries.csv` (12 rows): the rookie and 10-year-veteran minimum salary for each season.
- `team_payrolls.csv` (360 rows): total payroll per team-season. We only use it to verify our own payroll totals.

**Important metrics.**
- `salary`: a player's salary (cap hit) with a team for a season, in **US dollars (nominal, not inflation-adjusted)**. Values range from about \$5K (a few days on a 10-day contract) to about \$60M (supermax stars). The median is about \$2.7M, and the distribution is very right-skewed.
- `salary_cap`: the soft cap on team payroll for a season, in USD. It grew from \$70.0M in 2015-16 to \$154.6M in 2025-26 (the jump in 2016 came from a new TV deal). Because the cap more than doubled, **raw dollars aren't comparable across seasons**, so we convert salaries to **% of that season's cap**. A max contract *starts* at 25%, 30%, or 35% of the cap depending on years of service. Raises of up to 8% a year can push its later years to about 40% (e.g. Stephen Curry in 2021-22, after the COVID cap freeze), but nothing should come close to 45%.
- `luxury_tax_line`, `first_apron`, `second_apron` (USD): payroll thresholds above the cap. A team over the tax line pays a penalty tax on every dollar over. Since the 2023 CBA, teams over the first apron, and even more so the second apron, lose key trade tools (for example, they can't take back more salary than they send out, or combine salaries in trades). Before 2023-24 there was no second apron (it is blank), and `first_apron` holds the older "tax apron" (hard-cap line). We use these to place every team-season in a **cap tier**: under cap → over cap → over tax → over 1st apron → over 2nd apron. We expect teams in higher tiers to be under more pressure to trade salary away.
- `rookie_minimum` / `ten_year_veteran_minimum` (USD): the smallest full-season salary a player can earn. A salary below the rookie minimum means the player was not on a standard full-season contract (a 10-day deal, a two-way deal, a prorated signing, or being waived).
- `two_way`: True if the player was on a two-way contract (split between the NBA team and its G League affiliate, paid much less than a standard contract).
- Derived: `pct_of_team_payroll` (the player's share of their team's payroll) and `changed_team_next_season` (whether the player's main team changed in the following season, which can come from a trade *or* free agency).

**Concerns.**
1. *Snapshot, not final accounting.* The card warns that these are contract and cap-hit snapshots, which can differ from final figures (incentives, dead money, cap holds, converted contracts). The 2026-27 season is a forward-looking snapshot: only 483 rows, because many players hadn't signed yet. We exclude it from the cleaned tables.
2. *Contract details are mostly empty.* `player_option`, `qualifying_offer`, and `terminated` are False in every row, `team_option` is True in only 12 rows (all 2026-27), and `notes` is always blank. The dataset has no contract length or years remaining. That was one of the variables we wanted most for our hypothesis, so we will need another source for it (e.g. Spotrac or Basketball-Reference contract pages).
3. *Missing IDs are not missing at random.* 157 rows (2%) have no `player_id`. They are mostly fringe or low-salary players and rows where HoopsHype used a legal name instead of a common name (e.g. "Louis Williams" = Lou Williams, "Patrick Mills" = Patty Mills). Because of this we key players by a normalized name instead of the ID. Name variants could still split one player into two keys.
4. *Inconsistent `team_id`.* IDs are 1–29 for 29 teams, but Charlotte's is 5312. We drop `team_id` and identify teams by name.
5. *Multiple rows per player-season.* 439 player-seasons have salary with more than one team (traded or waived mid-season). Five player-seasons have two rows *with the same team* (e.g. a two-way deal converted to a standard one, or a waived contract plus a new one). These are real but need to be combined before we have one row per player-season.
6. *Possibly incomplete seasons.* 2019-20 has fewer rows (518) and smaller average rosters (~17 players per team vs. ~21 in other seasons), and no team is above the tax line that season. This suggests missing salary rows (COVID-era roster churn may not have been captured). We will cross-check 2019-20 payrolls against Basketball-Reference before relying on them.
7. *Source bias.* Salary data compiled by HoopsHype (a media site) may be most accurate for high-profile players and less careful for end-of-bench or 10-day players.

```python
## 5A. LOAD the dataset (raw CSVs downloaded into data/00-raw/ by the setup cell)
import re
import unicodedata

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

salaries_raw      = pd.read_csv('data/00-raw/hf_player_salaries.csv')
cap_raw           = pd.read_csv('data/00-raw/hf_cap_tax_aprons.csv')
min_raw           = pd.read_csv('data/00-raw/hf_min_salaries.csv')
team_payrolls_raw = pd.read_csv('data/00-raw/hf_team_payrolls.csv')


def normalize_name(name):
    """Name key used to match players across sources: no accents/punctuation/suffixes, lower case."""
    if not isinstance(name, str):
        return np.nan
    name = unicodedata.normalize('NFKD', name).encode('ascii', 'ignore').decode()
    name = re.sub(r"[.'`’]", '', name.lower())   # "P.J." -> "pj", "De'Aaron" -> "deaaron"
    name = re.sub(r'[-,]', ' ', name)
    name = re.sub(r'\b(jr|sr|ii|iii|iv|v)\b', ' ', name)
    return ' '.join(name.split())


display(salaries_raw.head())
display(cap_raw.head(3))
```

#### 5F. Clean the data (and justify missing-data handling)

```python
## 5F. CLEAN + build tidy analysis tables
# Cleaning decisions (and why):
#  1. Drop the 2026-27 season. It is an incomplete forward-looking snapshot (483 rows vs ~600-700),
#     and its offseason (summer 2027) has not happened yet.
#  2. Drop columns that carry no information: notes (100% missing), source (one constant value),
#     player_option / qualifying_offer / terminated (always False), team_option (only set in 2026-27),
#     and team_id (inconsistent; team name is a cleaner key).
#  3. Keep rows with a missing player_id (no dropna). These are real salaries that count toward team
#     payroll, and the missingness is concentrated among low-salary players and name variants, so dropping
#     them would bias payroll and roster numbers. Instead every player gets a normalized-name key
#     (`name_key`), which is also how we will match players to the transactions data (Dataset #6).
#  4. Keep low/partial salaries (below the rookie minimum) but flag them. They are real (10-day deals,
#     waived players, two-way contracts) and matter for payroll, but a model may want to exclude them.
#  5. Combine the 5 (season, player, team) pairs with two rows into one row by summing salary.
#     They are two contracts with the same team in the same season.
salaries = (salaries_raw[salaries_raw['season_year'] <= 2025]
            .drop(columns=['notes', 'source', 'team_option', 'player_option', 'qualifying_offer',
                           'terminated', 'team_id'])
            .assign(name_key=lambda d: d['player'].map(normalize_name)))

# Sanity check: does the name key ever merge two *different* player ids in one season?
collisions = (salaries.dropna(subset=['player_id'])
              .groupby(['season', 'name_key'])['player_id'].nunique())
print('Name-key collisions within a season (different ids, same key):', (collisions > 1).sum())

contracts = (salaries
             .groupby(['season_year', 'season', 'name_key', 'team'], as_index=False)
             .agg(player=('player', 'first'), player_id=('player_id', 'first'),
                  salary=('salary', 'sum'), two_way=('two_way', 'any'),
                  n_contract_rows=('salary', 'size')))

# Team-season table: payroll and cap tier
team_season = (contracts.groupby(['season_year', 'season', 'team'], as_index=False)
               .agg(payroll=('salary', 'sum'), n_players=('name_key', 'nunique'),
                    n_two_way=('two_way', 'sum'))
               .merge(cap_raw[['season', 'salary_cap', 'luxury_tax_line', 'first_apron', 'second_apron']],
                      on='season', how='left'))
team_season['payroll_pct_cap'] = (100 * team_season['payroll'] / team_season['salary_cap']).round(1)
team_season['cap_tier'] = np.select(
    [team_season['second_apron'].notna() & (team_season['payroll'] >= team_season['second_apron']),
     team_season['payroll'] >= team_season['first_apron'],
     team_season['payroll'] >= team_season['luxury_tax_line'],
     team_season['payroll'] >= team_season['salary_cap']],
    ['over_2nd_apron', 'over_1st_apron', 'over_tax', 'over_cap'], default='under_cap')

# Check our payroll totals against the dataset's own team_payrolls table
chk = team_season.merge(team_payrolls_raw[['season', 'team', 'payroll']], on=['season', 'team'],
                        suffixes=('', '_hf'))
print(f"Team payrolls matching team_payrolls.csv: {(chk['payroll'] == chk['payroll_hf']).mean():.0%} "
      f"of {len(chk)} team-seasons")

# Player-season table: one row per player per season. The primary team is the team that paid the most.
contracts = contracts.merge(team_season[['season', 'team', 'payroll']], on=['season', 'team'], how='left')
contracts['pct_of_team_payroll'] = 100 * contracts['salary'] / contracts['payroll']
player_season = (contracts.sort_values('salary', ascending=False)
                 .groupby(['season_year', 'season', 'name_key'], as_index=False)
                 .agg(player=('player', 'first'), player_id=('player_id', 'first'),
                      primary_team=('team', 'first'), total_salary=('salary', 'sum'),
                      pct_of_team_payroll=('pct_of_team_payroll', 'first'),
                      n_teams=('team', 'nunique'), two_way=('two_way', 'any'))
                 .merge(cap_raw[['season', 'salary_cap']], on='season', how='left')
                 .merge(min_raw[['season', 'rookie_minimum']], on='season', how='left'))
player_season['salary_pct_cap'] = 100 * player_season['total_salary'] / player_season['salary_cap']
player_season['flag_below_rookie_min'] = player_season['total_salary'] < player_season['rookie_minimum']
player_season = player_season.drop(columns=['salary_cap', 'rookie_minimum'])

# Next season's primary team. NaN if the player has no salary next season (left the league / unsigned)
# or if this is 2025-26 (we dropped 2026-27).
nxt = (player_season[['season_year', 'name_key', 'primary_team']]
       .assign(season_year=lambda d: d['season_year'] - 1)
       .rename(columns={'primary_team': 'next_season_primary_team'}))
player_season = player_season.merge(nxt, on=['season_year', 'name_key'], how='left')
player_season['changed_team_next_season'] = np.where(
    player_season['next_season_primary_team'].isna(), np.nan,
    (player_season['next_season_primary_team'] != player_season['primary_team']).astype(float))
player_season['salary_pct_cap'] = player_season['salary_pct_cap'].round(2)
player_season['pct_of_team_payroll'] = player_season['pct_of_team_payroll'].round(2)

print('player_season:', player_season.shape, '| team_season:', team_season.shape)
print('Duplicate (season, player) keys left:', player_season.duplicated(['season', 'name_key']).sum())
print('Missing values in player_season:'); display(player_season.isna().sum()[lambda s: s > 0].to_frame('n_missing').T)

# A few summary statistics relevant to the project
display(player_season[['total_salary', 'salary_pct_cap', 'pct_of_team_payroll']].describe().round(2))
display(pd.crosstab(team_season['season'], team_season['cap_tier'])
        .reindex(columns=['under_cap', 'over_cap', 'over_tax', 'over_1st_apron', 'over_2nd_apron'],
                 fill_value=0))
print('Share of player-seasons whose primary team changed the next season (trade OR free agency):',
      round(player_season['changed_team_next_season'].mean(), 3))
player_season.head()
```

### Dataset #6: NBA.com Official Player Movement Log (July 2015 – present)

**Source and access.** NBA.com's transactions page is built from a single public JSON file, `https://stats.nba.com/js/data/playermovement/NBA_Player_Movement.json`. It needs no account or API key, and the setup cell above downloads it in one request into `data/00-raw/nba_player_movement.json`. This is the league's own record of transactions, so it is more authoritative than fan-maintained sites. 

**Structure and important variables.** The file has one row per *transaction line*, about 10,000 rows from 2015-07-01 onward. Each row has:
- `Transaction_Type`: one of `Trade`, `Signing`, `Waive`, `ContractConverted` (e.g. two-way to standard), or `AwardOnWaivers` (claimed off waivers). Trades are what we want to predict. The other types help explain *why* a player changed teams (for example, a player who changed teams without being traded left in free agency).
- `TRANSACTION_DATE`: the date the transaction became official (a day, with no time). We use it to decide whether a trade happened during the **offseason** or the **season**. Trading is frozen from the trade deadline (February, or March in 2021) until the season ends, so there are no trades in April or May. We define the offseason after season *S*-(*S*+1) as **April 1 of year *S*+1 up to the day before opening night of the next season**. This also handles the 2020 bubble year, when the offseason ran from October to December 22, 2020.
- `TRANSACTION_DESCRIPTION`: free text such as *"Washington Wizards received guard Tyus Jones from Memphis Grizzlies."* For trades we extract the **receiving team**, the **sending team**, the player's **position** (guard / forward / center), and the **player name**. Picks and cash appear only as *"received draft consideration,"* with no detail about which pick.
- `TEAM_ID` / `TEAM_SLUG`: the receiving team (NBA.com's 10-digit team ID and a short name like `thunder`). `Additional_Sort`: for trades, the NBA team ID of the **sending** team; for other types it is 0.
- `PLAYER_ID` / `PLAYER_SLUG`: NBA.com's player ID and URL slug. `PLAYER_ID` is 0 for "draft consideration" rows.
- `GroupSort`: the transaction type plus an ID, e.g. `Trade 2022130`. All rows of one trade share this ID, so it tells us which players moved in the same deal. A trade has 1–16 rows (one per asset received by one team).

Each traded player becomes one row in `offseason_traded_players` (offseason year, player, from team, to team). That gives us the outcome label for the player-season table from Dataset #5: was this player traded in the offseason that followed this season?

**Concerns.**
1. *Coverage starts July 1, 2015.* Trades made during the 2015 draft (June 25, 2015) are missing, so the offseason after 2014-15 is incomplete. This doesn't affect us, because our salary data (Dataset #5) starts with 2015-16, whose offseason is summer 2016.
2. *Picks and cash have no detail.* Every pick or cash payment is recorded as "draft consideration" with `PLAYER_ID = 0`. We can tell a trade included picks, but not how many or how valuable they were.
3. *Missing values are coded as 0.* There are no NaN cells, but `PLAYER_ID = 0` and `Additional_Sort = 0` mean "not applicable." If we treated those zeros as real IDs, joins and counts would be wrong, so we convert them to NaN.
4. *Small inconsistencies.* The claimed-off-waivers type is spelled `AwardOnWaivers` in `Transaction_Type` but `AwardedOnWaivers` in `GroupSort`. The Clippers are "LA Clippers" here and "Los Angeles Clippers" in Dataset #5. Ten signings were recorded twice under different IDs.
5. *Name-based linking.* NBA.com player IDs differ from the HoopsHype IDs in Dataset #5, so we link the two on a normalized player name. Players with different names in the two sources (e.g. legal name vs. nickname) won't match, and newly drafted players traded on draft night have no earlier NBA salary to match to.
6. *Re-routed trades.* Some players are traded twice within days (e.g. Buddy Hield in September 2026, Atlanta → Charlotte → Chicago), where one team acts as a short stop in a larger deal. They count once in our offseason label, but for the second trade their "from" team is not the team they actually played for, so we use their *first* offseason trade.
7. *Live file.* The file is updated as new transactions happen, so re-downloading it later adds rows. We record the date range we loaded so the results can be reproduced.

#### 6B. Tidy the dataset (or show it is already tidy)

```python
## 6B. TIDY the dataset
# The raw table is *almost* tidy (one row per transaction line), but three columns pack several variables
# into one string, so we split them:
#   - GroupSort "Trade 2022130"  -> transaction type + transaction_id (the id that groups one trade's rows)
#   - TRANSACTION_DESCRIPTION    -> for trades: to_team, asset (position + player, or "draft consideration"),
#                                   from_team
#   - TRANSACTION_DATE (text)    -> a real datetime
# After this, each row is one observation (one asset received by one team in one transaction) and each
# column is one variable.
movement = movement_raw.rename(columns={
    'Transaction_Type': 'transaction_type', 'TRANSACTION_DESCRIPTION': 'description',
    'TEAM_ID': 'team_id', 'TEAM_SLUG': 'team_slug', 'PLAYER_ID': 'player_id',
    'PLAYER_SLUG': 'player_slug', 'Additional_Sort': 'from_team_id'})
movement['date'] = pd.to_datetime(movement['TRANSACTION_DATE'])
movement['transaction_id'] = movement['GroupSort'].str.split(' ').str[-1]

TRADE_PATTERN = (r'^(?P<to_team>.+?) received '
                 r'(?:(?P<position>guard|forward|center) (?P<player>.+?)|(?P<other_asset>draft consideration))'
                 r' from (?P<from_team>.+?)\.?$')
is_trade = movement['transaction_type'].eq('Trade')
parsed = movement.loc[is_trade, 'description'].str.extract(TRADE_PATTERN)
movement = movement.join(parsed)

print('Trade rows whose description did NOT fit the pattern:',
      parsed['to_team'].isna().sum(), 'of', is_trade.sum())
movement = movement.drop(columns=['TRANSACTION_DATE', 'GroupSort'])
display(movement[is_trade].head())
```

```python
## 6C. SIZE of the dataset
trades_only = movement[movement['transaction_type'].eq('Trade')]
print('All transaction rows:', movement.shape)
print(f"Date range: {movement['date'].min().date()} to {movement['date'].max().date()}")
print('Trade rows:', len(trades_only), '| distinct trades:', trades_only['transaction_id'].nunique(),
      '| distinct traded players:', trades_only['player_id'].replace(0, np.nan).nunique())
display(movement['transaction_type'].value_counts().rename('rows').to_frame().T)

# Rows by calendar year and type (2015 starts in July; the current year is still in progress)
display(pd.crosstab(movement['date'].dt.year, movement['transaction_type']).T)

# When do trades happen? No trades in April-May (trading is frozen after the deadline until the season
# ends), which is what lets us separate offseason trades from in-season trades.
fig, ax = plt.subplots(figsize=(9, 3))
trades_only['date'].dt.month.value_counts().reindex(range(1, 13), fill_value=0).plot.bar(ax=ax)
ax.set_title('Trade rows by calendar month (2015-2026)')
ax.set_xlabel('Month'); ax.set_ylabel('Trade rows')
plt.tight_layout(); plt.show()
```

#### 6D. Missing data: how much, where, and is it random?

```python
## 6D. MISSINGNESS -- how much, where, and is it random?
# 1) Explicit NaNs in the raw file: none
display(movement_raw.isna().sum().rename('n_NaN').to_frame().T)

# 2) Missing values hidden as 0 (sentinel values)
sentinel = pd.DataFrame({
    'player_id == 0':    movement['player_id'].eq(0),
    'from_team_id == 0': movement['from_team_id'].eq(0)})
display(sentinel.groupby(movement['transaction_type']).sum())
# -> player_id is 0 only for trade rows that move "draft consideration" (picks/cash), and from_team_id is
#    0 for every non-trade row. Both are STRUCTURAL ("not applicable"), not random. We convert them to NaN
#    in 6F so they don't get treated as real ids.
print('Trade rows with player_id 0 that are NOT draft consideration:',
      (movement['player_id'].eq(0) & is_trade & movement['other_asset'].isna()).sum())

# 3) Missing by design in the parsed trade columns: position/player are empty exactly when the asset is
#    draft consideration, and the other way around.
display(movement.loc[is_trade, ['position', 'player', 'other_asset']].isna().sum().rename('n_missing').to_frame().T)

# 4) Missing coverage (whole periods absent): the file begins 2015-07-01, so 2015 draft-night trades
#    (June 25, 2015) are missing. Rows per year are much lower in 2015 because only half the year is covered.
print('First trade in the file:', trades_only['date'].min().date())
```

#### 6E. Outliers and suspicious entries

```python
## 6E. OUTLIERS / SUSPICIOUS ENTRIES
trades_only = movement[is_trade]

# Trade size: rows (assets received) per trade, flagged with the IQR rule
rows_per_trade = trades_only.groupby('transaction_id').size()
q1, q3 = rows_per_trade.quantile([.25, .75])
big_cut = q3 + 1.5 * (q3 - q1)
print(f'Rows per trade: median {rows_per_trade.median():.0f}, max {rows_per_trade.max()}; '
      f'IQR outlier cut-off > {big_cut:.1f} -> {(rows_per_trade > big_cut).sum()} large trades flagged')
biggest = rows_per_trade.idxmax()
display(trades_only[trades_only['transaction_id'] == biggest][['date', 'description']])

suspicious = pd.DataFrame({
    # exact same date + description + player recorded twice under different ids
    'duplicate_entry': movement.duplicated(['date', 'description', 'player_id'], keep=False),
    # a team trading with itself would be impossible
    'trade_with_self': is_trade & movement['to_team'].eq(movement['from_team']),
    'trade_in_apr_may': is_trade & movement['date'].dt.month.isin([4, 5]),   # trading is frozen then
    # parsed player name should agree with NBA.com's own slug
    'name_slug_mismatch': is_trade & movement['player'].notna()
                          & (movement['player'].str.lower().str.replace(r'[^a-z0-9 -]', '', regex=True)
                             .str.replace(' ', '-') != movement['player_slug']),
    'future_date': movement['date'] > pd.Timestamp.today().normalize(),
})
display(suspicious.sum().rename('n_flagged').to_frame().T)
display(movement[suspicious['duplicate_entry']].sort_values(['date', 'description'])
        [['date', 'transaction_type', 'description', 'transaction_id']].head(6))

# Each team name in the text should always map to the same team id (and vice versa)
print('Receiving-team names per TEAM_ID (should all be 1):',
      trades_only.groupby('team_id')['to_team'].nunique().max())
print('Sending-team names per from_team_id (should all be 1):',
      trades_only.groupby('from_team_id')['from_team'].nunique().max())

# Players traded more than once within a few weeks (re-routed through a third team)
player_trades = trades_only.dropna(subset=['player']).sort_values('date')
gap = player_trades.groupby('player_id')['date'].diff().dt.days
quick = player_trades[gap.le(14) | gap.shift(-1).le(14)]
print('Trade rows where the same player is traded again within 14 days:', len(quick))
display(quick[['date', 'description']].head(6))

# Label typo: two spellings for the same transaction type
print('Type spellings:', sorted(movement['transaction_type'].unique()),
      '| GroupSort prefixes:', sorted(movement_raw['GroupSort'].str.split(' ').str[0].unique()))
```

#### 6F. Clean the data (and justify missing-data handling)

```python
## 6F. CLEAN + build the offseason-trade table
# Cleaning decisions (and why):
#  1. Sentinel zeros -> NaN (player_id 0 and from_team_id 0 mean "not applicable"). We do NOT drop these
#     rows: draft-consideration rows are real parts of trades (they tell us a trade included picks/cash).
#  2. Drop the 10 duplicated signing rows (same date, description and player under two ids): keep the first.
#  3. Harmonize the waiver-claim label ("AwardOnWaivers" vs "AwardedOnWaivers" -> "AwardedOnWaivers").
#  4. Map NBA.com team names to the names used in Dataset #5 ("LA Clippers" -> "Los Angeles Clippers").
#  5. Assign each transaction to an NBA season and phase (offseason vs in-season) using opening-night
#     dates. Offseason after season S = April 1 of S+1 up to the day before the next opening night.
#  6. Large trades and quickly re-routed players are kept (they are real), but stay flagged.
movement_clean = (movement
                  .drop_duplicates(['date', 'description', 'player_id'], keep='first')
                  .assign(player_id=lambda d: d['player_id'].replace(0, np.nan),
                          from_team_id=lambda d: d['from_team_id'].replace(0, np.nan),
                          transaction_type=lambda d: d['transaction_type']
                                                      .replace('AwardOnWaivers', 'AwardedOnWaivers'))
                  .reset_index(drop=True))
for col in ['to_team', 'from_team']:
    movement_clean[col] = movement_clean[col].replace('LA Clippers', 'Los Angeles Clippers')

# Opening night of each season (season_year = first calendar year of the season, as in Dataset #5).
# Source: NBA schedules; 2020-21 opened late (Dec 22, 2020) after the COVID bubble.
OPENING_NIGHT = pd.to_datetime(pd.Series({
    2015: '2015-10-27', 2016: '2016-10-25', 2017: '2017-10-17', 2018: '2018-10-16', 2019: '2019-10-22',
    2020: '2020-12-22', 2021: '2021-10-19', 2022: '2022-10-18', 2023: '2023-10-24', 2024: '2024-10-22',
    2025: '2025-10-21', 2026: '2026-10-20'}))


def season_and_phase(date):
    """Return (season_year whose games come before this date or contain it, phase)."""
    started = OPENING_NIGHT[OPENING_NIGHT <= date]
    season = started.index.max() if len(started) else 2014          # before Oct 2015 -> 2014-15 season
    if date >= pd.Timestamp(year=season + 1, month=4, day=1):
        return season, 'offseason'      # after this season's trading freeze, before the next opener
    return season, 'in_season'


movement_clean[['season_year', 'phase']] = movement_clean['date'].apply(
    lambda d: pd.Series(season_and_phase(d)))
movement_clean['season_year'] = movement_clean['season_year'].astype(int)
# The offseason after season S belongs to S: e.g. a trade on 2025-07-06 is in the offseason after 2024-25.
movement_clean.loc[movement_clean['phase'].eq('offseason'), 'offseason_after_season'] = \
    movement_clean['season_year']

trades_clean = movement_clean[movement_clean['transaction_type'].eq('Trade')].copy()
trades_clean['large_trade_flag'] = trades_clean['transaction_id'].map(rows_per_trade > big_cut)
display(pd.crosstab(trades_clean['season_year'], trades_clean['phase']))

# One row per player traded in a given offseason. If a player was traded more than once in the same
# offseason (re-routed), keep the first trade, whose from_team is the team the player actually played for.
offseason_traded_players = (trades_clean[trades_clean['phase'].eq('offseason') & trades_clean['player'].notna()]
                            .sort_values('date')
                            .assign(name_key=lambda d: d['player'].map(normalize_name),
                                    n_offseason_trades=lambda d: d.groupby(['offseason_after_season', 'player_id'])
                                                                  ['transaction_id'].transform('nunique'))
                            .drop_duplicates(['offseason_after_season', 'player_id'], keep='first')
                            [['offseason_after_season', 'date', 'transaction_id', 'player_id', 'player',
                              'name_key', 'position', 'from_team', 'to_team', 'n_offseason_trades']]
                            .astype({'offseason_after_season': int, 'player_id': int})
                            .reset_index(drop=True))

print('Rows removed (duplicates):', len(movement) - len(movement_clean))
print('Remaining NaNs are structural (not-applicable fields):')
display(trades_clean[['player_id', 'player', 'position', 'other_asset', 'from_team_id']].isna().sum().to_frame('n_NaN').T)
display(offseason_traded_players.groupby('offseason_after_season').size().rename('players traded').to_frame().T)
offseason_traded_players.head()
```
